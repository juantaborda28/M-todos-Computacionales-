```
#armonico simple
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

#datos
m = float(input("Ingrese la masa m (kg): "))
k = float(input("Ingrese la constante del resorte k (N/m): "))

x0 = float(input("Ingrese la posición inicial x0 (m): "))
v0 = float(input("Ingrese la velocidad inicial v0 (m/s): "))

#sistema de ecuaciones
def sistema_amortiguado(t, y):
    x, v = y
    dxdt = v
    dvdt = - (k/m)*x
    return [dxdt, dvdt]

#condiciones iniciales
y0 = [x0, v0]

#tiempo
t_span = (0, 200)
t_eval = np.linspace(0, 200, 1000)

# solucion de la edo
sol = solve_ivp(sistema_amortiguado, t_span, y0, t_eval=t_eval)

#grafica
plt.figure(figsize=(10, 5))
plt.plot(sol.t, sol.y[0], label="x(t) Posición", color="blue")
plt.xlabel("Tiempo (s)")
plt.ylabel("Posición (m)")
plt.title("Sistema masa–resorte armonico simple")
plt.grid()
plt.legend()
plt.show()
```

```
#amortiguado
#datos
m = float(input("Ingrese la masa m (kg): "))
k = float(input("Ingrese la constante del resorte k (N/m): "))
c = float(input("Ingrese el coeficiente de amortiguamiento c (kg/s): "))

x0 = float(input("Ingrese la posición inicial x0 (m): "))
v0 = float(input("Ingrese la velocidad inicial v0 (m/s): "))

#sistema de ecuaciones
def sistema_amortiguado(t, y):
    x, v = y
    dxdt = v
    dvdt = -(c/m)*v - (k/m)*x
    return [dxdt, dvdt]

#condiciones iniciales
y0 = [x0, v0]

#tiempo
t_span = (0, 200)
t_eval = np.linspace(0, 200, 1000)

# solucion de la edo
sol = solve_ivp(sistema_amortiguado, t_span, y0, t_eval=t_eval)

#grafica
plt.figure(figsize=(10, 5))
plt.plot(sol.t, sol.y[0], label="x(t) Posición", color="blue")
plt.xlabel("Tiempo (s)")
plt.ylabel("Posición (m)")
plt.title("Sistema masa–resorte amortiguado")
plt.grid()
plt.legend()
plt.show()
```

```
#energía amortiguado

# Datos de entrada
m = float(input("Ingrese la masa m (kg): "))
k = float(input("Ingrese la constante del resorte k (N/m): "))
c = float(input("Ingrese el coeficiente de amortiguamiento c (kg/s): "))

x0 = float(input("Ingrese la posición inicial x0 (m): "))
v0 = float(input("Ingrese la velocidad inicial v0 (m/s): "))

# Sistema de ecuaciones diferenciales
def sistema_forzado(t, y):
    x, v = y
    dxdt = v
    dvdt = -(c/m)*v - (k * x) / m
    return [dxdt, dvdt]

# Condiciones iniciales
y0 = [x0, v0]

# Tiempo
t_span = (0, 20)
t_eval = np.linspace(0, 20, 1000)

# Solución de la EDO
sol = solve_ivp(sistema_forzado, t_span, y0, t_eval=t_eval)

# Energía cinética y potencial
E_c = 0.5 * m * sol.y[1]**2  # Energía cinética
E_p = 0.5 * k * sol.y[0]**2  # Energía potencial

# Energía total (sin considerar disipación)
E_total = E_c + E_p


# Graficar las energías
plt.figure(figsize=(10, 6))
plt.plot(sol.t, E_c, label="Energía Cinética", color="blue")
plt.plot(sol.t, E_p, label="Energía Potencial", color="green")
plt.plot(sol.t, E_total, label="Energía Total", color="red")

plt.xlabel("Tiempo (s)")
plt.ylabel("Energía (J)")
plt.title("Energía en un sistema masa–resorte amortiguado")
plt.legend()
plt.grid(True)
plt.show()
```

```
#forzado
#datos
m = float(input("Ingrese la masa m (kg): "))
k = float(input("Ingrese la constante del resorte k (N/m): "))
c = float(input("Ingrese el coeficiente de amortiguamiento c (kg/s): "))

F0 = float(input("Ingrese la amplitud de la fuerza F0 (N): "))
omega = float(input("Ingrese la frecuencia angular ω (rad/s): "))

x0 = float(input("Ingrese la posición inicial x0 (m): "))
v0 = float(input("Ingrese la velocidad inicial v0 (m/s): "))

# sistema de ecuaciones
def sistema_forzado(t, y):
    x, v = y
    dxdt = v
    dvdt = (F0*np.cos((omega*t)) - c*v - k*x) / m
    return [dxdt, dvdt]

#condiciones iniciales
y0 = [x0, v0]

#tiempo
t_span = (0, 200)
t_eval = np.linspace(0, 200, 1000)

#solución de la edo
sol = solve_ivp(sistema_forzado, t_span, y0, t_eval=t_eval)

#grafica del movimiento
plt.figure(figsize=(10, 5))
plt.plot(sol.t, sol.y[0], label="x(t) Posición", color="purple")
plt.xlabel("Tiempo (s)")
plt.ylabel("Posición (m)")
plt.title("Sistema masa–resorte amortiguado y forzado")
plt.grid()
plt.legend()
plt.show()
```

```
#velocidad y aceleración forzado
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Datos de entrada
m = float(input("Ingrese la masa m (kg): "))
k = float(input("Ingrese la constante del resorte k (N/m): "))
c = float(input("Ingrese el coeficiente de amortiguamiento c (kg/s): "))

F0 = float(input("Ingrese la amplitud de la fuerza F0 (N): "))
omega = float(input("Ingrese la frecuencia angular ω (rad/s): "))

x0 = float(input("Ingrese la posición inicial x0 (m): "))
v0 = float(input("Ingrese la velocidad inicial v0 (m/s): "))

# Sistema de ecuaciones diferenciales
def sistema_forzado(t, y):
    x, v = y
    dxdt = v
    dvdt = (F0 * np.cos(omega * t) - c * v - k * x) / m
    return [dxdt, dvdt]

y0 = [x0, v0]

# Tiempo
t_span = (0, 200)
t_eval = np.linspace(0, 200, 1000)


sol = solve_ivp(sistema_forzado, t_span, y0, t_eval=t_eval)

v = sol.y[1]

a = (F0 * np.cos(omega * sol.t) - c * v - k * sol.y[0]) / m

# Graficar Velocidad y Aceleraciónplt.figure(figsize=(10, 6))

# Graficar velocidad
plt.subplot(2, 1, 1)
plt.plot(sol.t, v, label="Velocidad (m/s)", color="blue")
plt.xlabel("Tiempo (s)")
plt.ylabel("Velocidad (m/s)")
plt.title("Velocidad del Sistema")
plt.grid(True)
plt.legend()

plt.subplot(2, 1, 2)
plt.plot(sol.t, a, label="Aceleración (m/s²)", color="red")
plt.xlabel("Tiempo (s)")
plt.ylabel("Aceleración (m/s²)")
plt.title("Aceleración del Sistema")
plt.grid(True)
plt.legend()

# Mostrar los gráficos
plt.tight_layout()
plt.show()
```
