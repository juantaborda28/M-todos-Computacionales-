```python
#1A
import numpy as np

def bransdon(A, b, tol=1e-6, max_iter=1000):

    n = A.shape[0]
    x = np.zeros(n)
    x_new = np.copy(x)

    D = np.diag(A)
    if np.any(D == 0):
        raise ValueError("La matriz A tiene ceros en la diagonal")
    b_norm = np.linalg.norm(b)
    for k in range(max_iter):
        for i in range(n):
            sum_val = 0
            for j in range(n):
                if i != j:
                    sum_val += A[i, j] * x[j]
            x_new[i] = (b[i] - sum_val) / D[i]

        residual_norm = np.linalg.norm(b - A @ x_new)

        if b_norm == 0: #b es un vector cero
            if residual_norm < tol:
                return x_new, k + 1
        elif residual_norm / b_norm < tol:
            return x_new, k + 1

        x = np.copy(x_new) # Actualizar x para la siguiente iteración

```

```python
#1B
A = np.array([[1, -1, 2, 0],
              [-2, 11, 0, -1],
              [3, -1, 10, -1],
              [0, 2, -1, 8]])
b = np.array([6, 25, -12, 14])

x, iterations = bransdon(A, b)

print("Solución x:", x)
print("Iteraciones:", iterations)
```

```python
#1C
x_exact = np.linalg.solve(A, b)
print("Solución exacta x_exact:", x_exact)

```

```python
#2ByC
import numpy as np
import matplotlib.pyplot as plt
m = 1.0
k = 4.0
c = 0.5
def f(t, y):
    y1, y2 = y
    dy1_dt = y2
    dy2_dt = -(k/m) * y1 - (c/m) * y2
    return np.array([dy1_dt, dy2_dt])

y1_0 = 1
y2_0 = 0
y_inicial = np.array([y1_0, y2_0])

t_1 = 0
t_final = 20
n_steps = 100
dt = (t_final - t_1) / n_steps
t_val = np.linspace(t_1, t_final, n_steps + 1)

sol_y = np.zeros((n_steps + 1, 2))
sol_y[0] = y_inicial
# metodo heun
for i in range(n_steps):
    current_t = t_val[i]
    current_y = sol_y[i]

    k1 = f(current_t, current_y)
    y_predictor = current_y + dt * k1

    k2 = f(current_t + dt, y_predictor)

    sol_y[i+1] = current_y + (dt/2) * (k1 + k2)

y1_solucion = sol_y[:, 0]

plt.figure(figsize=(10, 6))
plt.plot(t_val, y1_solucion, label='desplazamiento y(t)')
plt.title('Solución de sistema masa-resorte')
plt.xlabel('Tiempo (t)')
plt.ylabel('Desplazamiento y(t)')
plt.grid(True)
plt.legend()
plt.show()
```
