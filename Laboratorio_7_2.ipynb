```python
#actividad 1
import numpy as np
from scipy import integrate

# EJEMPLO DE FUNCIÓN f(x)
def f(x):
    return np.sin(x)

a = 0
b = np.pi

def trapecio(f, a, b, n):
    x = np.linspace(a, b, n+1)
    y = f(x)
    h = (b - a) / n
    return h * (0.5*y[0] + np.sum(y[1:-1]) + 0.5*y[-1])

n = 1000
T_manual = trapecio(f, a, b, n)

x = np.linspace(a, b, n+1)
y = f(x)
T_scipy = integrate.trapezoid(y, x)

T_quad, _ = integrate.quad(lambda x: f(x), a, b)

print("Método del Trapecio (manual):", T_manual)
print("SciPy trapezoid:", T_scipy)
print("integrate.quad:", T_quad)
```

```python
#actividad 2
import numpy as np
from scipy import integrate
def trapecio(f, a, b, n):
    x = np.linspace(a, b, n+1)
    y = f(x)
    h = (b - a) / n
    return h * (0.5*y[0] + np.sum(y[1:-1]) + 0.5*y[-1])
def simpson(f, a, b, n):
    if n % 2 != 0:
        n += 1
    x = np.linspace(a, b, n+1)
    y = f(x)
    h = (b - a) / n

    return (h/3) * (y[0] + y[-1] +
                    4*np.sum(y[1:-1:2]) +
                    2*np.sum(y[2:-2:2]))
```

```python
#integral 1
f1 = lambda x: np.sqrt(1 + x)
a, b = 0, 0.1
n = 50
T1 = trapecio(f1, a, b, n)
S1 = simpson(f1, a, b, n)
Real1, _ = integrate.quad(f1, a, b)
print("Trapecio:", T1)
print("Simpson:", S1)
print("Valor real:", Real1)
print()
```

```python
#Actividad 3
def trapecio_compuesto(f, a, b, n):
    x = np.linspace(a, b, n+1)
    y = f(x)
    h = (b - a) / n
    return h * (0.5*y[0] + np.sum(y[1:-1]) + 0.5*y[-1])
def simpson_compuesto(f, a, b, n):
    if n % 2 != 0:
        n += 1
    x = np.linspace(a, b, n+1)
    y = f(x)
    h = (b - a) / n
    return (h/3) * (y[0] + y[-1] +
                    4*np.sum(y[1:-1:2]) +
                    2*np.sum(y[2:-2:2]))
f = lambda x: 1 + np.cos(x)**2 + x
a = -0.5
b = 1.5
n = 100
T = trapecio_compuesto(f, a, b, n)
S = simpson_compuesto(f, a, b, n)
Real, _ = integrate.quad(f, a, b)

print("RESULTADOS ACTIVIDAD 3\n")
print("Trapecio compuesto:", T)
print("Simpson compuesto:", S)
print("Valor real (quad):", Real)
print("\nErrores:")
print("Error Trapecio:", abs(Real - T))
print("Error Simpson:", abs(Real - S))
```
